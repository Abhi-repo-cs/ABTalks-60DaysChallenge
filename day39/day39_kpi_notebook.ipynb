# Day 39 — Business KPI Analytics
Build a KPI monitoring system for customer, revenue, and retention performance.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('business_kpi_data.csv', parse_dates=['month'])
df.tail()
```

## KPI Definitions
- Customer growth = month-over-month change in total customers
- Revenue growth = month-over-month change in revenue
- Retention = 1 − churn rate
- ARPU = revenue / active customers
- AOV = revenue / orders


```python
latest = df.iloc[-1]
print('Customers:', f"{latest.customers:,.0f}")
print('Revenue:', f"${latest.revenue:,.0f}")
print('Retention:', f"{latest.retention_rate:.1%}")
print('ARPU:', f"${latest.arpu:.2f}")
print('AOV:', f"${latest.avg_order_value:.2f}")
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax.plot(df.month, df.revenue)
ax.set_title('Revenue Trend')
ax.set_xlabel('Month'); ax.set_ylabel('Revenue ($)')
plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax.plot(df.month, df.retention_rate * 100)
ax.set_title('Retention Rate Trend')
ax.set_xlabel('Month'); ax.set_ylabel('Retention (%)')
plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
